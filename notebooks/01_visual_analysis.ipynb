```python
sessions_orders = sessions.merge(
    orders[['order_id', 'order_status']],
    on='order_id',
    how='left'
)
```

```python
sessions_orders.head()
```

```python
sessions_orders['is_completed'] = (
    sessions_orders['order_status'] == 'completed'
).astype(int)
```

```python
sessions_orders[['order_id', 'order_status', 'is_completed']].head(10)
```

```python
sessions_orders['is_completed'].sum()
```

```python
device_cr = (
    sessions_orders
    .groupby(['month', 'device_type'])
    .agg(
        sessions=('session_id', 'count'),
        completed_orders=('is_completed', 'sum')
    )
    .reset_index()
)
```

```python
device_cr['conversion_rate'] = (
    device_cr['completed_orders'] / device_cr['sessions'] * 100
).round(2)
```

```python
device_cr[
    device_cr['month'].astype(str).isin(['2024-06', '2024-07', '2024-08'])
]
```

```python
device_cr_recent = device_cr[
    device_cr['month'].astype(str).isin(['2024-06', '2024-07', '2024-08'])
].copy()
```

```python
plt.figure(figsize=(10, 5))

for device in device_cr_recent['device_type'].unique():
    data = device_cr_recent[device_cr_recent['device_type'] == device]

    plt.plot(
        data['month'].astype(str),
        data['conversion_rate'],
        marker='o',
        label=device
    )

plt.title('Conversion Rate by Device')
plt.xlabel('Month')
plt.ylabel('Conversion Rate, %')
plt.legend()

plt.savefig('../outputs/04_conversion_rate_by_device.png', dpi=150, bbox_inches='tight')

plt.show()
```

```python
mobile_funnel = (
    sessions_orders[
        (sessions_orders['device_type'] == 'mobile') &
        (sessions_orders['month'].astype(str).isin(['2024-06', '2024-07', '2024-08']))
    ]
    .groupby('month')
    .agg(
        sessions=('session_id', 'count'),
        add_to_cart=('add_to_cart', 'sum'),
        checkout_started=('checkout_started', 'sum'),
        completed_orders=('is_completed', 'sum')
    )
    .reset_index()
)
```

```python
mobile_funnel['session_to_cart'] = (
    mobile_funnel['add_to_cart'] / mobile_funnel['sessions'] * 100
).round(2)

mobile_funnel['cart_to_checkout'] = (
    mobile_funnel['checkout_started'] / mobile_funnel['add_to_cart'] * 100
).round(2)

mobile_funnel['checkout_to_purchase'] = (
    mobile_funnel['completed_orders'] / mobile_funnel['checkout_started'] * 100
).round(2)
```

```python
mobile_funnel
```

```python
plt.figure(figsize=(10, 5))

plt.plot(
    mobile_funnel['month'].astype(str),
    mobile_funnel['session_to_cart'],
    marker='o',
    label='Session → Cart'
)

plt.plot(
    mobile_funnel['month'].astype(str),
    mobile_funnel['cart_to_checkout'],
    marker='o',
    label='Cart → Checkout'
)

plt.plot(
    mobile_funnel['month'].astype(str),
    mobile_funnel['checkout_to_purchase'],
    marker='o',
    label='Checkout → Purchase'
)

plt.title('Mobile Funnel Conversion')
plt.xlabel('Month')
plt.ylabel('Conversion Rate, %')
plt.legend()

plt.savefig('../outputs/05_mobile_funnel.png', dpi=150, bbox_inches='tight')

plt.show()
```

```python
payments['payment_datetime'] = pd.to_datetime(payments['payment_datetime'])
```

```python
payments.info()
```

```python
payments_analysis = payments.merge(
    sessions_orders[['order_id', 'device_type', 'month']],
    on='order_id',
    how='left'
)
```

```python
payments_analysis.head()
```

```python
payments_analysis['is_failed'] = (
    payments_analysis['payment_status'] == 'failed'
).astype(int)
```

```python
payments_analysis[['payment_status', 'is_failed']].head(10)
```

```python
payments_analysis['is_failed'].sum()
```

```python
device_failures = (
    payments_analysis[
        payments_analysis['month'].astype(str).isin(['2024-06', '2024-07', '2024-08'])
    ]
    .groupby(['month', 'device_type'])
    .agg(
        payment_attempts=('payment_id', 'count'),
        failed_payments=('is_failed', 'sum')
    )
    .reset_index()
)
```

```python
device_failures['failure_rate'] = (
    device_failures['failed_payments'] / device_failures['payment_attempts'] * 100
).round(2)
```

```python
device_failures
```

```python
device_failures_main = device_failures[
    device_failures['device_type'].isin(['desktop', 'mobile'])
].copy()

plt.figure(figsize=(10, 5))

for device in device_failures_main['device_type'].unique():
    data = device_failures_main[
        device_failures_main['device_type'] == device
    ]

    plt.plot(
        data['month'].astype(str),
        data['failure_rate'],
        marker='o',
        label=device
    )

plt.title('Payment Failure Rate by Device')
plt.xlabel('Month')
plt.ylabel('Failure Rate, %')
plt.legend()

plt.savefig('../outputs/06_payment_failure_rate.png', dpi=150, bbox_inches='tight')

plt.show()
```

```python
mobile_payment_methods = (
    payments_analysis[
        (payments_analysis['device_type'] == 'mobile') &
        (payments_analysis['month'].astype(str).isin(['2024-06', '2024-07', '2024-08']))
    ]
    .groupby(['month', 'payment_method'])
    .agg(
        payment_attempts=('payment_id', 'count'),
        failed_payments=('is_failed', 'sum')
    )
    .reset_index()
)

mobile_payment_methods['failure_rate'] = (
    mobile_payment_methods['failed_payments'] /
    mobile_payment_methods['payment_attempts'] * 100
).round(2)

mobile_payment_methods
```

```python
mobile_card_failures = (
    payments_analysis[
        (payments_analysis['device_type'] == 'mobile') &
        (payments_analysis['payment_method'] == 'Банковская карта') &
        (payments_analysis['payment_status'] == 'failed') &
        (payments_analysis['month'].astype(str).isin(['2024-06', '2024-07', '2024-08']))
    ]
    .groupby(['month', 'failure_reason'])
    .agg(
        failed_payments=('payment_id', 'count')
    )
    .reset_index()
)

mobile_card_failures
```

### Вывод

Падение mobile conversion связано преимущественно с этапом завершения покупки.

Payment failure rate на mobile вырос с 9,88% в июне до 28,39% в августе. Основной рост ошибок приходится на оплату банковской картой: failure rate увеличился с 10,67% до 48,39%.

Главной новой причиной стала ошибка `3-D Secure / тайм-аут`: в июне такие ошибки отсутствовали, в июле зафиксировано 46 случаев, в августе — 65.

Это указывает на отдельную техническую проблему mobile card payment flow.

# Marketing Analysis: VK Ads

```python
vk_campaigns = (
    sessions_orders[
        (sessions_orders['channel'] == 'VK Ads') &
        (sessions_orders['month'].astype(str).isin(['2024-06', '2024-07', '2024-08']))
    ]
    .groupby(['month', 'campaign_id'])
    .agg(
        sessions=('session_id', 'count'),
        add_to_cart=('add_to_cart', 'sum'),
        completed_orders=('is_completed', 'sum')
    )
    .reset_index()
)
```

```python
vk_campaigns['cart_rate'] = (
    vk_campaigns['add_to_cart'] / vk_campaigns['sessions'] * 100
).round(2)

vk_campaigns['conversion_rate'] = (
    vk_campaigns['completed_orders'] / vk_campaigns['sessions'] * 100
).round(2)
```

```python
vk_campaigns
```

```python
plt.figure(figsize=(10, 5))

for campaign in vk_campaigns['campaign_id'].unique():
    data = vk_campaigns[
        vk_campaigns['campaign_id'] == campaign
    ]

    plt.plot(
        data['month'].astype(str),
        data['conversion_rate'],
        marker='o',
        label=campaign
    )

plt.title('VK Ads Conversion Rate by Campaign')
plt.xlabel('Month')
plt.ylabel('Conversion Rate, %')
plt.legend()

plt.savefig('../outputs/07_vk_campaign_cr.png', dpi=150, bbox_inches='tight')

plt.show()
```

### Вывод

Кампания VK_SUMMER привела большой объём трафика, однако его качество существенно ниже VK_RETARGET.

Conversion Rate VK_SUMMER составил 2,12% в июле и 1,83% в августе против 4,35% и 5,64% у VK_RETARGET.

Разница проявляется уже на этапе Session → Cart, поэтому низкую эффективность кампании нельзя объяснить только технической проблемой оплаты.

При этом VK_SUMMER значительно хуже и по экономике:
- CPA: 7 048 ₽ в июле и 8 371 ₽ в августе;
- ROAS: 2,77 и 3,67.

Для VK_RETARGET за те же месяцы:
- CPA: 2 453 ₽ и 1 980 ₽;
- ROAS: 12,82 и 18,00.

Таким образом, масштабирование VK_SUMMER увеличило трафик, но ухудшило его качество и эффективность рекламных расходов.

```python
items_analysis = items_products.merge(
    orders[['order_id', 'order_datetime', 'order_status', 'month']],
    on='order_id',
    how='left'
)
```

```python
items_analysis.head()
```

```python
completed_items_recent = items_analysis[
    (items_analysis['order_status'] == 'completed') &
    (items_analysis['month'].astype(str).isin(['2024-06', '2024-07', '2024-08']))
].copy()
```

```python
completed_items_recent[['month', 'category', 'quantity', 'unit_price', 'line_amount']].head()
```

```python
category_sales = (
    completed_items_recent
    .groupby(['month', 'category'])
    .agg(
        items_sold=('quantity', 'sum'),
        product_sales=('line_amount', 'sum'),
        avg_unit_price=('unit_price', 'mean')
    )
    .reset_index()
)

category_sales
```

```python
selected_categories = [
    'Смартфоны',
    'Компьютеры и планшеты',
    'Смарт-часы и браслеты',
    'Аксессуары'
]

product_mix = category_sales[
    category_sales['category'].isin(selected_categories)
].pivot(
    index='category',
    columns='month',
    values='items_sold'
)

product_mix
```

```python
product_mix.plot(
    kind='bar',
    figsize=(12, 6)
)

plt.title('Product Mix by Category')
plt.xlabel('Category')
plt.ylabel('Items Sold')
plt.xticks(rotation=20)
plt.legend(title='Month')

plt.savefig('../outputs/09_product_mix.png', dpi=150, bbox_inches='tight')

plt.show()
```

### Вывод

Снижение Average Order Value после июня связано с изменением product mix.

Количество проданных смартфонов снизилось с 213 единиц в июне до 172 в июле и 181 в августе. Продажи компьютеров и планшетов снизились со 102 до 75 и 72 единиц соответственно.

При этом продажи более дешёвых аксессуаров остались относительно стабильными: 265 единиц в июне, 246 в июле и 265 в августе.

Среднее количество товаров в completed-заказе практически не изменилось, поэтому снижение AOV объясняется не уменьшением размера корзины, а снижением доли дорогих товаров в структуре продаж.
